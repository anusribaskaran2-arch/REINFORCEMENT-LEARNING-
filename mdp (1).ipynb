# Task 1 & 2: Environment Tracking & Discovery

```python
import gymnasium as gym

# 1. Total Number of Environments
envs = list(gym.envs.registry.keys())
print(f"Total number of environments: {len(envs)}")

# 2. Print All Environment Names
print("\n--- First 20 Registered Environments ---")
for name in sorted(envs)[:20]:
    print(name)
```

## Task 3: Space Exploration of Benchmark Environments


```python
environments = [
    "CartPole-v1",
    "FrozenLake-v1",
    "MountainCar-v0",
    "Blackjack-v1",
    "Taxi-v4",       # Updated to modern version syntax
    "CliffWalking-v1" # Updated to modern version syntax
]

for env_name in environments:
    print(f"\n{'='*20} Exploring: {env_name} {'='*20}")
    try:
        env = gym.make(env_name)
        print(f"Action Space:      {env.action_space}")
        print(f"Observation Space: {env.observation_space}")
        env.close()
    except Exception as e:
        print(f"Could not load {env_name}: {e}")
```
