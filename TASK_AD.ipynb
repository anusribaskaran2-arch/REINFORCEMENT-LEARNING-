```python
import numpy as np
import pandas as pd

# 1. Load the click-through dataset
# Assuming the file path is 'ads_dataset.csv'
df = pd.read_csv('Ads_CTR_Optimisation (1).csv')

# 2. Set Multi-Armed Bandit Parameters
epsilon = 0.1  # 10% Exploration, 90% Exploitation
num_ads = df.shape[1]
num_users = df.shape[0]

# Track variables
ad_counts = np.zeros(num_ads)        # Times each ad was shown (N_i)
ad_rewards = np.zeros(num_ads)       # Total successful clicks per ad (R_i)
ad_estimated_ctr = np.zeros(num_ads) # Average CTR per ad (Q_i)

# Seed for reproducibility
np.random.seed(42)

# 3. Run the Simulation Loop
for user_idx in range(num_users):
    # Determine whether to explore or exploit
    if np.random.rand() < epsilon:
        # Explore: pick a completely random ad
        selected_ad = np.random.randint(num_ads)
    else:
        # Exploit: pick the ad with the current highest estimated CTR
        # If there's a tie, choose randomly among the best
        max_ctr = np.max(ad_estimated_ctr)
        best_ads = np.where(ad_estimated_ctr == max_ctr)[0]
        selected_ad = np.random.choice(best_ads)

    # Get user interaction from data (1 if clicked, 0 if not)
    reward = df.iloc[user_idx, selected_ad]

    # 4. Update memory metrics
    ad_counts[selected_ad] += 1
    ad_rewards[selected_ad] += reward
    ad_estimated_ctr[selected_ad] = ad_rewards[selected_ad] / ad_counts[selected_ad]

# 5. Determine the winner
best_ad_index = np.argmax(ad_estimated_ctr)
best_ad_name = df.columns[best_ad_index]

print(f"Simulation Finished across {num_users} users.")
print(f"Best Advertisement: {best_ad_name} with an estimated CTR of {ad_estimated_ctr[best_ad_index]:.4f}")
```
