```python
#UNIT V: DATA ANALYSIS

#Experiment 1: Filtering
#Aim: To perform row filtering in a DataFrame based on specific conditions.
#Simple
#Q: Create a DataFrame of students and filter those who scored more than 80 in Math.
#Medium
#Q: Filter employees in the “HR” department with salary > 50,000.
#Q: Filter transactions after Jan 1, 2023, amount > ₹10,000, and made via Credit Card.
#Code:
# Simple
import pandas as pd
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'], 'Math': [85, 78, 92]})
print(df[df['Math'] > 80])

# Medium
df = pd.DataFrame({'Name': ['John', 'Sara', 'Mike'],
                   'Department': ['HR', 'IT', 'HR'],
                   'Salary': [60000, 48000, 52000]})
print(df[(df['Department'] == 'HR') & (df['Salary'] > 50000)])

# Application-Based
df = pd.DataFrame({
    'Date': pd.to_datetime(['2023-01-01', '2023-01-10', '2023-01-20']),
    'Amount': [5000, 15000, 20000],
    'Payment_Mode': ['Cash', 'Credit Card', 'Credit Card']
})
print(df[(df['Date'] > '2023-01-01') & (df['Amount'] > 10000) & (df['Payment_Mode'] == 'Credit Card')])

#Experiment 2: Subsetting Data
#Aim: To subset specific rows and columns from a DataFrame.
#Simple
#Q: Select only the Name and Age columns.
#Medium
#Q: Get rows where English > 70, only return Name and English.
#Application-Based
#Q: Subset patients > 60 years with columns Patient_ID, Age, Diagnosis, and Medication.
#Code:
# Simple
df = pd.DataFrame({'Name': ['A', 'B'], 'Age': [21, 22], 'Gender': ['M', 'F']})
print(df[['Name', 'Age']])

# Medium
df = pd.DataFrame({'Name': ['A', 'B', 'C'], 'English': [72, 65, 85], 'Math': [80, 70, 90]})
print(df[df['English'] > 70][['Name', 'English']])

# Application-Based
df = pd.DataFrame({
    'Patient_ID': [101, 102, 103],
    'Age': [65, 45, 72],
    'Diagnosis': ['Diabetes', 'Hypertension', 'Arthritis'],
    'Medication': ['DrugA', 'DrugB', 'DrugC']
})
print(df[df['Age'] > 60][['Patient_ID', 'Age', 'Diagnosis', 'Medication']])

#Experiment 3: Grouping the Data
#Aim: To group and summarize DataFrame data using groupby and aggregation.
#Simple
#Q: Group by Department and count employees.
#Medium
#Q: Group by Region and Product, find total revenue.
#Application-Based
#Q: Group bank data by Branch and Account_Type, find avg/min/count.
#Code:
# Simple
import pandas as pd
import numpy as np
df = pd.DataFrame({'Name': ['A', 'B', 'C'], 'Department': ['HR', 'IT', 'HR']})
print(df.groupby('Department')['Name'].count())

# Medium
df = pd.DataFrame({
    'Region': ['North', 'South', 'North', 'South'],
    'Product': ['X', 'Y', 'X', 'Y'],
    'Revenue': [1000, 2000, 1500, 2500]
})
print(df.groupby(['Region', 'Product'])['Revenue'].sum())

# Application-Based
df = pd.DataFrame({
    'Branch': ['A', 'A', 'B', 'B', 'B'],
    'Account_Type': ['Saving', 'Saving', 'Saving', 'Current', 'Saving'],
    'Balance': [30000, 25000, 40000, 20000, 35000]
})
print(df.groupby(['Branch', 'Account_Type']).agg(
    Avg_Balance=('Balance', 'mean'),
    Min_Balance=('Balance', 'min'),
    Count=('Balance', 'count')
))

#Experiment 4: Aggregating Data
#Aim: To perform basic aggregation operations on Pandas DataFrames.
#Simple
#Q: Get mean, min, max of a Sales column.
#Medium
#Q: Group scores by Subject and get mean, max, min.
#Application-Based
#Q: Analyze electricity usage per household.
#Code:
# Simple
df = pd.DataFrame({'Sales': [100, 200, 300, 400]})
print("Mean:", df['Sales'].mean())
print("Min:", df['Sales'].min())
print("Max:", df['Sales'].max())

# Medium
df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D'],
    'Subject': ['Math', 'Math', 'English', 'English'],
    'Score': [85, 90, 78, 88]
})
print(df.groupby('Subject').agg({'Score': ['mean', 'max', 'min']}))

# Application-Based
df = pd.DataFrame({
    'Household': ['H1', 'H1', 'H2', 'H2', 'H3'],
    'Units': [10, 20, 30, 40, 25]
})
grouped = df.groupby('Household')['Units']
summary = grouped.agg(Total='sum', Average='mean')
summary['Max_Usage'] = summary['Total'].max() == summary['Total']
print(summary)

#Experiment 5: Time Series Data
#Aim: To manipulate and analyze time series data using Pandas.
#Simple
#Q: Convert string dates to datetime and sort.
#Medium
#Q: Resample hourly data to daily average.
#Application-Based
#Q: Analyze stock prices for Jan 2023.
#Code:
# Simple
df = pd.DataFrame({'Date': ['2023-01-03', '2023-01-01', '2023-01-02']})
df['Date'] = pd.to_datetime(df['Date'])
df.sort_values('Date', inplace=True)
print(df)

# Medium
import numpy as np
df = pd.DataFrame({
    'Datetime': pd.date_range('2023-01-01', periods=48, freq='H'),
    'Temperature': np.random.randint(20, 35, size=48)
})
df.set_index('Datetime', inplace=True)
print(df.resample('D').mean())

# Application-Based
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', periods=31),
    'Company': ['ABC']*31,
    'Closing_Price': np.random.randint(100, 200, 31)
})
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
jan_df = df.loc['2023-01-01':'2023-01-31']
print(jan_df.groupby('Company')['Closing_Price'].mean())

# Plot trend
jan_df['Closing_Price'].plot(title='ABC Stock Price - Jan 2023')
plt.show()


```
